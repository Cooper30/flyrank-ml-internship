## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 pages. Among them, 16,262 pages (54.2%) show a declining trend, which suggests that there are many potential candidates for review. Using a simple initial rule of at least 180 days since the last update and at least 500 impressions in the last 90 days, I found 17 pages that are both stale and still visible. These numbers suggest that a prioritization approach could be useful: rather than reviewing thousands of declining pages manually, the goal is to identify a smaller set of pages that deserve attention first.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/Cooper30/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)

declining_pages = (df["trend_direction"] == "down").sum()
declining_pct = declining_pages / total_pages * 100

stale_visible = (
    (df["days_since_last_update"] >= 180) &
    (df["impressions_90d"] >= 500)
).sum()

print(f"Total pages in starter dataset: {total_pages:,}")
print(f"Pages with a declining trend: {declining_pages:,} ({declining_pct:.1f}%)")
print(f"Stale but visible pages: {stale_visible:,}")
```
